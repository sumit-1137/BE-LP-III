```python
""""Implement Gradient Descent Algorithm to find the local minima of a function.
For example, find the local minima of the function y=(x+3)² starting from the point x=2."""
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def function(x):
    return (x + 3)**2

def derivative(x):
    return 2 * (x + 3)
```

```python
x = 2

learning_rate = 0.1

iterations = 100

print("Initial X:", x)
```

```python
for i in range(iterations):

    gradient = derivative(x)

    x = x - learning_rate * gradient

    print("Iteration:", i + 1, "X:", round(x, 6))
```

```python
print("Local Minimum X:", round(x, 6))

print("Minimum Value Y:", round(function(x), 6))
```

```python
x_values = np.linspace(-6, 3, 100)

y_values = function(x_values)

plt.plot(x_values, y_values)

plt.scatter(-3, 0, color="red")

plt.xlabel("X")
plt.ylabel("Y")

plt.title("Gradient Descent")

plt.grid()
plt.show()
```
